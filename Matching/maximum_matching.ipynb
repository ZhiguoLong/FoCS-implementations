```python
from networkx.algorithms import bipartite
import networkx as nx
import matplotlib.pyplot as plt
```

```python
def draw_matching_graph(B,M,visited,a=None):
    X = B[0]
    Y = B[1]
    edges = B[2]
    G = nx.Graph()
    G.add_nodes_from(X, bipartite=0)
    G.add_nodes_from(Y, bipartite=1)
    G.add_edges_from(edges)
    pos = nx.bipartite_layout(G, X,align='horizontal')
    nx.draw(G, pos, with_labels=True)
    nx.draw_networkx_edges(G, pos, edgelist=M, edge_color='r', width=4)
    nx.draw_networkx_nodes(G,pos,nodelist=visited,node_color='g')
    if a:
        nx.draw_networkx_nodes(G,pos,nodelist=[a],node_color='y')
    plt.show()
```

```python
# B: [X,Y,edges]
# M: {u:v, v:u, ...}
def breadth_first_search_scratch(B,M):
    visited = set()
    Ux = set()
    pred = dict()
    Y = B[1]
    X = B[0]
    edges = B[3]
    
    M_tuple = [(x,M[x]) for x in X if x in M]
    print("X:",X)
    print("Y:",Y)
    for x in X:
        if x not in M:
            Ux.add(x)
    print("Ux:",Ux)
    L = list(Ux)
    print("L:",L)
    for l in L:
        visited.add(l)
    # print("visited:",visited)
    while len(L) > 0:
        a = L.pop(0)
        print("a:",a)
        print("L:",L)
        if a in X:
            for b in edges[a]:
                print("b:",b)
                if b not in visited and (a not in M or b != M[a]):
                    visited.add(b)
                    draw_matching_graph(B,M_tuple,visited,a)
                    pred[b] = a
                    if b in M:
                        L.append(b)
                    else:
                        P = [b]
                        c = b
                        while c not in Ux:
                            d = pred[c]
                            c = d
                            P.append(c)
                        P.reverse()
                        return P
        else:
            b = M[a]
            visited.add(b)
            draw_matching_graph(B,M_tuple,visited,a)
            pred[b] = a
            L.append(b)
        print(L)
        
```

```python
def maximum_matching(B,M={}):
    P = breadth_first_search_scratch(B,M)
    while P is not None:
        # adjust M
        m = len(P)
        for i in range(m-1):
            if i % 2==0:
                x = P[i]
                y = P[i+1]
                M[x] = y 
                M[y] = x
            else:
                x = P[i]
                y = P[i+1]
                M.pop(x)
                M.pop(y)
        P = breadth_first_search_scratch(B,M)
    return M
```
